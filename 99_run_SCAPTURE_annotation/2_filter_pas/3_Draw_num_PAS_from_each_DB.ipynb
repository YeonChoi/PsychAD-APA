```python
import subprocess as sp
import pathlib
from collections import Counter

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
base_path = pathlib.Path('/sc/arion/projects/CommonMind/yeon/p/APA/')
```

```python
ANNO_BED_FILES = {
    'GENCODE': base_path / 'ref/GENCODE.v38.ensembl.104.nochr/gencode.v38.polyAs.polyA_site.nochr.bed.gz',
    'PolyA_DB3': base_path / 'ref/polyADB3/polyADB_hg19.human.PAS.nochr.hg38.bed.gz',
    'PolyA-Seq': base_path / 'ref/polyA-Seq/polyA-Seq_merged_hg38.nochr.bed.gz',
    'PolyASite2': base_path / 'ref/polyASite/atlas.clusters.2.0.GRCh38.96.bed.gz',
    'PolyASite3': base_path / 'ref/polyASite/atlas.clusters.3.0.GRCh38.GENCODE_42.Blood_Brain_RPM0.1.nochr.bed.gz',
}
```

```python
line_counts = {}
for key, fp in ANNO_BED_FILES.items():
    output = sp.check_output(f'zcat {fp} | wc -l', shell=True)
    line_counts[key] = int(output.decode().strip())
```

```python
fig, ax = plt.subplots(figsize=(3, 2))
pd.Series(line_counts).iloc[::-1].plot(ax=ax, kind='barh', width=0.75)

ticks = np.array([0, 200, 400, 600, 800])
realticks = ticks * 1000
ax.set_xticks(realticks)
ax.set_xticklabels([f'{t:d}K' for t in ticks])
ax.set_ylabel('PAS database')
ax.set_xlabel('Number of annotated PAS')

plt.savefig('Number_of_PAS_in_each_DB.pdf', bbox_inches='tight')
plt.show()
```
